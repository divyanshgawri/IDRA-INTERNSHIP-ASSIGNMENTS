```python
# Student Management System
# Data Structure: List of Dictionaries
students = []

def add_student():
    """Adds a new student to the system."""
    print("\n--- Add New Student ---")
    student_id = input("Enter Student ID: ").strip()

    # Check if ID already exists
    for student in students:
        if student['ID'] == student_id:
            print("❌ Error: Student with this ID already exists!")
            return

    name = input("Enter Name: ").strip().title()
    course = input("Enter Course: ").strip().title()

    # Input validation for Age and Marks
    try:
        age = int(input("Enter Age: "))
        marks = float(input("Enter Marks: "))
    except ValueError:
        print("❌ Error: Age must be a whole number and Marks must be a number. Student not added.")
        return

    # Create dictionary and add to list
    student_record = {
        'ID': student_id,
        'Name': name,
        'Age': age,
        'Course': course,
        'Marks': marks
    }
    students.append(student_record)
    print(f"✅ Student '{name}' added successfully!")

def view_students():
    """Displays all student records."""
    print("\n--- All Student Records ---")
    if not students:
        print("No students found. The list is empty.")
        return

    # Print table header
    print(f"{'ID':<10} | {'Name':<20} | {'Age':<5} | {'Course':<15} | {'Marks':<5}")
    print("-" * 65)

    # Print student rows
    for student in students:
        print(f"{student['ID']:<10} | {student['Name']:<20} | {student['Age']:<5} | {student['Course']:<15} | {student['Marks']:<5.2f}")

def search_student():
    """Searches for a student by ID or Name."""
    print("\n--- Search Student ---")
    search_term = input("Enter Student ID or Name to search: ").strip().lower()

    found = False
    for student in students:
        # Check if the search term matches the ID or is part of the Name
        if search_term == student['ID'].lower() or search_term in student['Name'].lower():
            if not found:
                print("\n--- Search Results ---")
            print(f"ID: {student['ID']}, Name: {student['Name']}, Age: {student['Age']}, Course: {student['Course']}, Marks: {student['Marks']}")
            found = True

    if not found:
        print("❌ No matching student found.")

def update_student():
    """Updates an existing student's details."""
    print("\n--- Update Student ---")
    student_id = input("Enter the ID of the student to update: ").strip()

    for student in students:
        if student['ID'] == student_id:
            print(f"Student found: {student['Name']}")
            print("(Press Enter to keep the current value)")

            new_name = input(f"New Name [{student['Name']}]: ").strip().title()
            if new_name:
                student['Name'] = new_name

            new_course = input(f"New Course [{student['Course']}]: ").strip().title()
            if new_course:
                student['Course'] = new_course

            try:
                new_age = input(f"New Age [{student['Age']}]: ").strip()
                if new_age:
                    student['Age'] = int(new_age)

                new_marks = input(f"New Marks [{student['Marks']}]: ").strip()
                if new_marks:
                    student['Marks'] = float(new_marks)
            except ValueError:
                print("❌ Invalid number entered! Age/Marks were not updated.")

            print("✅ Student details updated successfully!")
            return

    print("❌ Student ID not found.")

def delete_student():
    """Deletes a student record by ID."""
    print("\n--- Delete Student ---")
    student_id = input("Enter the ID of the student to delete: ").strip()

    for i in range(len(students)):
        if students[i]['ID'] == student_id:
            deleted_name = students[i]['Name']
            del students[i]
            print(f"✅ Student '{deleted_name}' has been deleted.")
            return

    print("❌ Student ID not found.")

def main():
    """Main function to run the menu loop."""
    while True:
        print("\n" + "="*35)
        print("     STUDENT MANAGEMENT SYSTEM")
        print("="*35)
        print("1. Add New Student")
        print("2. View All Students")
        print("3. Search for a Student")
        print("4. Update Student Details")
        print("5. Delete a Student")
        print("6. Exit")
        print("="*35)

        choice = input("Enter your choice (1-6): ").strip()

        if choice == '1':
            add_student()
        elif choice == '2':
            view_students()
        elif choice == '3':
            search_student()
        elif choice == '4':
            update_student()
        elif choice == '5':
            delete_student()
        elif choice == '6':
            print("\nExiting the Student Management System. Goodbye!")
            break
        else:
            print("❌ Invalid choice! Please select a number between 1 and 6.")


if __name__ == "__main__":
    main()
```
